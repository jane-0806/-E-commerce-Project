# Clean the data

```python
# copy of the original data
clean_df = df.copy()
```

```python
# starting rows 
starting_rows = len(clean_df)
starting_rows
```

```python
# create a cleaning log for recording 
cleaning_log = []

cleaning_log.append({
    "Step":"Initial data",
    "Rows_Before":len(clean_df),
    "Rows_After": len(clean_df),
    "Rows_Removed": 0
})
```

```python
# Remove exact duplicates
rows_before_du = len(clean_df)
```

```python
clean_df = clean_df.drop_duplicates().copy()
```

```python
rows_after_du = len(clean_df)
```

```python
rows_removed_du = (rows_before_du - rows_after_du)

rows_removed_du
```

```python
cleaning_log.append({
    "Step":"Removed exact duplicates",
    "Rows_Before":rows_before_du,
    "Rows_After": rows_after_du,
    "Rows_Removed": rows_removed_du
})
```

```python
cleaning_log
```

```python
clean_df.shape
```

```python
#convert the InvoiceNo to string (text)
clean_df["InvoiceNo"]=(clean_df["InvoiceNo"].astype("string"))
```

```python
#convert the StockCode to string (text)
clean_df["StockCode"]=(clean_df["StockCode"].astype("string"))
```

```python
# Convert the CustomerID from decimal to integer, then text
clean_df["CustomerID"] = (
    clean_df["CustomerID"].astype("Int64").astype("string")
)
```

```python
# Ensure the data is a datetime 
clean_df["InvoiceDate"] = pd.to_datetime(
    clean_df["InvoiceDate"],
    errors="coerce"
)
```

```python
#cehck the data types of the columns
clean_df.dtypes
```

```python
# organise the space in the "description" column
clean_df["Description"] = (
    clean_df["Description"]
    .astype("string")
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

```python
# organise the space in the "country" column
clean_df["Country"] = (
    clean_df["Country"]
    .astype("string")
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

## Create transaction status 

```python
clean_df["TransactionStatus"] = "Completed Sales"
```

```python
# identify missing values in the InvoiceNo column
cancelled_mask = (
    clean_df["InvoiceNo"].str.startswith("C", na=False)
)
```

```python
# identify cancelled transactions for updating the TransactionStatus column
clean_df.loc[
    cancelled_mask,
    "TransactionStatus"
] = "Cancelled"
```

```python
# identify negative quantities without C
stock_adjustment_mask = (
    (clean_df["Quantity"]<0) & ~cancelled_mask
)
```

```python
# identify stock adjustment transactions for updating the TransactionStatus column
clean_df.loc[
    stock_adjustment_mask,
    "TransactionStatus"
] = "Stock Adjustment"
```

```python
#identify zero price rows 
zero_price_mask = clean_df["UnitPrice"] == 0
```

```python
#identify zero price transactions for updating the TransactionStatus column
clean_df.loc[
    zero_price_mask,
    "TransactionStatus"
] = "Price Adjustment"
```

```python
#identify negative price rows
negative_price_mask = clean_df["UnitPrice"] < 0
```

```python
# identify negative price transactions for updating the TransactionStatus column
clean_df.loc[
    negative_price_mask,
    "TransactionStatus"
] = "Price Adjustment"
```

```python
# cehck the TransactionStatus 
clean_df["TransactionStatus"].value_counts(
    dropna=False
)
```
