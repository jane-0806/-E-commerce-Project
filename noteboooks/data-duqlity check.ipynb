### Data Dictionary
- InvoiceNo (object): Unique idetifier for an invoice or transaction
- StockCode (object): Unique identifier for a product 
- Description (object): Name or decription of the product 
- Quantity (int64): Number of product units in the tansaction
- InvoiceData (detertime64): data and time of the transaction
- UniPrice (float64): Price of one product unit
- CustomerID (float64): Identifier for the customer
- Country (object): Country where the customer is located

```python
# Negative quantities 
negative_quantity = (df["Quantity"] < 0).sum()
negative_quantity
```

```python
df[df['Quantity']<0].head(10)
```

```python
# Zero quantities
zero_quantity = (df['Quantity']==0).sum()
zero_quantity
```

```python
# Negative Prices 
negative_price=(df['UnitPrice']<0).sum()
negative_price
```

```python
df[df["UnitPrice"] < 0].head(10)
```

```python
#zero price = free product 
zero_price = (df["UnitPrice"] == 0).sum()
zero_price
```

```python
df[df["UnitPrice"] == 0].head(10)
```

```python
# cancellation 
cancelled_invoice = (
    df["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
    .sum()
)

cancelled_invoice
```

```python
df[
    df["InvoiceNo"].astype(str).str.startswith("C")
].head(10)
```

```python
quality_counts = pd.Series({
    "Exact duplicate rows": df.duplicated().sum(),
    "Negative quantity rows": negative_quantity,
    "Zero quantity rows": zero_quantity,
    "Negative price rows": negative_price,
    "Zero price rows": zero_price,
    "Cancelled invoice rows": cancelled_invoice
})

quality_counts
```
