```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

### Monthly revenue and order

```python
#read the data
monthly_sales = pd.read_csv('D:\Project\online+retail\Yearmonth_revenue and order.csv')
```

```python
x = monthly_sales['Yearmonth']
y = monthly_sales['Monthly_revenue']

plt.figure(figsize=(12,5))

plt.plot(x,y)
plt.title("Monthly Revenue")
plt.ylabel('Revenue')
plt.xlabel('Year-Month')

plt.xticks(rotation=45, ha='right')
plt.ticklabel_format(axis='y', style='plain')
plt.tight_layout()

plt.show
```

```python
x = monthly_sales['Yearmonth']
y = monthly_sales['Monthly_order']

plt.figure(figsize=(12,5))

plt.plot(x,y)
plt.title("Monthly Order")
plt.ylabel('Orders')
plt.xlabel('Year_Month')

plt.xticks(rotation=45, ha='right')
plt.ticklabel_format(axis='y', style='plain')
plt.tight_layout()

plt.show()
```

### Weekday orders

```python
weekday_sales = pd.read_csv('D:\Project\online+retail\Weekday_order.csv')
```

```python
days = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

weekday_sales = (
    weekday_sales
    .set_index('Week_day')
    .reindex(days, fill_value=0)
    .reset_index()
)
```

```python
x = weekday_sales['Week_day']
y = weekday_sales['Weekday_order']

plt.figure(figsize=(12,5))

plt.bar(x,y)
plt.title("Weekday Orders")
plt.ylabel('Orders')
plt.xlabel('Weekday')

plt.xticks(rotation=45, ha='right')
plt.ticklabel_format(axis='y',style='plain')
plt.tight_layout()

plt.show()
```

### Revenue by country

```python
country_revenue = pd.read_csv('D:\Project\online+retail\Revenue by country.csv')
```

```python
x = country_revenue['Country']
y = country_revenue['total_revenue']

plt.figure(figsize=(12,5))

plt.bar(x,y)
plt.title("Revenue by Country")
plt.ylabel('Revenue')
plt.xlabel('Country')

plt.xticks(rotation=45, ha='right')
plt.ticklabel_format(axis='y',style='plain')
plt.tight_layout()

plt.show()
```

```python
other_countries = country_revenue[
    country_revenue['Country'] != 'United Kingdom'
].copy()
```

```python
top10_other_countries = (
    other_countries
    .sort_values('total_revenue', ascending=False)
    .head(10)
)
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    top10_other_countries['Country'],
    top10_other_countries['total_revenue']
)

plt.gca().invert_yaxis()

plt.title('Top 10 Countries by Revenue — Excluding the UK')
plt.xlabel('Revenue')
plt.ylabel('Country')
plt.ticklabel_format(axis='x', style='plain')

plt.tight_layout()
plt.show()
```

### number of customer by country

```python
country_customers = pd.read_csv(
    r'D:\Project\online+retail\number of customer by country.csv'
)
```

```python
x = country_customers['Country']
y = country_customers['Unique_customers']

plt.figure(figsize=(12,5))

plt.bar(x,y)
plt.title("Customer Count by Country")
plt.ylabel('Customer Count')
plt.xlabel('Country')

plt.xticks(rotation=45, ha='right')
plt.ticklabel_format(axis='y',style='plain')
plt.tight_layout()

plt.show()
```

```python
top10_customers = (
    country_customers[
        country_customers['Country'] != 'United Kingdom'
    ]
    .sort_values('Unique_customers', ascending=False)
    .head(10)
)
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    top10_customers['Country'],
    top10_customers['Unique_customers']
)

plt.gca().invert_yaxis()

plt.title('Top 10 Countries by Unique Customers — Excluding the UK')
plt.xlabel('Number of unique customers')
plt.ylabel('Country')

plt.tight_layout()
plt.show()
```
